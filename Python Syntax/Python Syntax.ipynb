- Reading, Writing and Appending modes

| Mode | Description |
|------|-------------|
| `r`  | Read text file only. File must already exist. |
| `rb` | Read binary file only (e.g., images, PDFs). File must already exist. |
| `r+` | Read and write text file. File must already exist. File pointer starts at the beginning. |
| `w`  | Write text file only. Creates the file if it does not exist. Overwrites existing content. |
| `wb` | Write binary file only. Creates the file if it does not exist. Overwrites existing content. |
| `w+` | Read and write text file. Creates the file if it does not exist. Overwrites existing content. |
| `a`  | Append text to the end of the file. Creates the file if it does not exist. Existing content is preserved. |
| `ab` | Append binary data to the end of the file. Creates the file if it does not exist. Existing content is preserved. |
| `a+` | Read and append text file. Creates the file if it does not exist. Writing always occurs at the end of the file. |
| `x`  | Create a new text file for writing. Raises `FileExistsError` if the file already exists. |
| `xb` | Create a new binary file for writing. Raises `FileExistsError` if the file already exists. |
| `x+` | Create a new text file for reading and writing. Raises `FileExistsError` if the file already exists. |

### Positional and Keyword Arguments

There are two kind of arguments, namely, keyword and positional. As the name suggests, the keyword argument is identified by a function based on some key whereas the positional argument is identified based on its position in the function definition. Let us have a look at this with an example.

**positional argument follows keyword argument** Syntax Error

```python
# this is signature of the function where we pass arguments

def foo(a, b, c=10): 
    print('a =', a) 
    print('b =', b) 
    print('c =', c) 
    
foo(2, 3)
foo(a=2, c=3, b=10)
foo(a=2, c=3, 9)
```

- In the first function call, we provided 2 arguments, but still the output is shown because of we provided 2 positional argument and the function has a default value for the final argument c. So, it takes the default value into account for the final argument.
- In the second function call, three keyword arguments are provided. The benefit of providing this keyword argument is that you need not remember the positions but just the keywords that are required for the function call. These keywords can be provided in any order but function will take these as key-value pairs and not in the order which they are being passed.
- In third function call, the error occurred because of the way we have passed the arguments during the function call. The error positional argument follows keyword argument means that the if any keyword argument is used in the function call then it should always be followed by keyword arguments. Positional arguments can be written in the beginning before any keyword argument is passed. Here, a=2 and c=3 are keyword argument. The 3rd argument 9 is a positional argument. This can not be interpreted by the python as to which key holds what value. The way python works in this regards is that, it will first map the positional argument and then any keyword argument if present.

### LEGB Rule for scope assignment


- L:Local -Names assigned in a any way within the function(def or lambda) and not declared global in that function
- E:Enclosing function local - names in the local scope of any and all enclosing functions(def or lambda), from inner to outer
- G:Gloabl -Names assigned at the top level of module or declared global in a def within the file
- B:Built-in -Names preassigned in built-in modules - open, rang etc.,

```python
# __len__() and __str__()

class Book:
    
    def __init__(self,author,title,pages):
        self.author = author
        self.title = title
        self.pages = pages
    
    def __str__(self):                                                        # this dunder methods returns string representation of class object
        return 'book author {} and title {}'.format(self.author, self.title)
    
    def __len__(self):
        return self.pages
    
    
my_book = Book('Prasoon','Python basics', 200)
print(my_book)                                                                # print function wants string represntation of my_book object     
len(my_book)
```

### Class/instance/static method, class/instance variables

| Method Type | Decorator | First Argument | Access Instance Data | Access Class Data | Typical Use |
|------------|------------|---------------|---------------------|------------------|-------------|
| Instance Method | None | `self` | Yes | Yes | Operate on object-specific data |
| Class Method | `@classmethod` | `cls` | No | Yes | Work with class-level data or alternative constructors |
| Static Method | `@staticmethod` | None | No | No | Utility/helper functions related to the class, simiarlar non-public method |


  **Instance Method**
- instance methods are varibles are called with 'self' inside class and with instance name outside class
  self.msg(), s2.msg(),self.name, s2.name

**Class Method**
- A class method is a method that is bound to the class and not the object of the class.
- They have the access to the state of the class as it takes keyword 'cls' in argument that points to the class and not the object instance(doesn't take 'self' in argument)
- It can modify a class state that would apply across all the instances of the class. For example, it can modify a class variable that will be applicable to all the instances.
- We use @classmethod decorator in python to create a class method
- use the keyword 'cls' as contrast to 'self' in instance method

**Static Method**
- A static method does not receive an implicit first argument like self or cls and therefore it  cant access or modify class & instance variables. 
- A static method is also a method that is bound to the class and not the object of the class. 
- This method can’t access or modify the class state. It is present in a class because it makes sense for the method to be present in class.
- we use @staticmethod decorator to create a static method in python.
- these methods are used as utility/helper functions.
- Helper functions similar to single underscore Non-public method

```python

import pandas as pd

class MarketDataProcessor:

    default_date_col = "trade_date"                                    # Class variable

    def __init__(self, input_df: pd.DataFrame):
        self.df = input_df                                             # Instance variable

    # Instance Method
    def calculate_returns(self, price_col: str) -> pd.DataFrame:
        self.df["returns"] = self.df[price_col].pct_change()
        return self.df

    # Class Method
    @classmethod
    def from_csv(cls, file_path: str):
        input_df = pd.read_csv(file_path)
        return cls(input_df)

    # Static Method
    @staticmethod
    def validate_columns(input_df: pd.DataFrame, required_cols: list) -> bool:
        return all(col in input_df.columns for col in required_cols)
    
```

### Generators

```python
# you create generator function which creats generator object( also called iterator) which can hold sequence of values
# iterables like string, list,tuple,dictionary
# Generators are iterators, a kind of iterable you can only iterate over once using loops
# Generators do not store all the values in memory, they generate the values on the fly, need to cast them in list

def get_fibonaci(n):                           # generator function
    a=1
    b=1
    
    for i in range(n):
        a,b=b,a+b
        yield a
        
my_generator = get_fibonaci(10)                 # this creates generator object
my_generator                                    # shows memory position of generator object   
next(my_generator)                              # next() works for generator but not for iterables

for i in my_generator:
    print(i)

s = 'hello'
s_iter = iter(s)                                # iter()- to convert iterable to generator
```

### Decorators

```python
# decortors can add extra functionality to existing function (also an object) without modifying its structure
# here, original function is passed as an argument for decorator function
# original func can only add ints, now with decorator function it can add tuple elements


def decorator_func(f):
    def wrapper(list_of_tuples):
        return [f(val[0], val[1]) for val in list_of_tuples]
    return wrapper


@decorator_func
def original_func(a, b):
    return a + b

original_func([(1,2),(3,4),(5,6)])
```

### Working with XML files

```python
# XML creates a tree-like structure that is easy to interpret and supports a hierarchy
# The largest, top-level element is called the root, which contains all other elements.Elements can contain markup, including other elements, which are called "child elements."
# XML documents have sections called elements, which are defined by a beginning and an ending tag
# Attributes are name–value pair that exist within a start-tag or empty-element tag. An XML attribute can only have a single value and each attribute can appear at most once on each element.

import xml.etree.ElementTree as ET

# The root variable is assumed to be an instance of an ElementTree object, which represents the entire XML document
tree = ET.parse("C:/Admin/C Drive/Python Syntax/xml_file.xml")                     # ET.parse() - ElementTree object
root = tree.getroot()                                                              # tree.getroot()

# information about root element
root.tag                                                                           # root.tag - tag of root element
root.attrib                                                                        # root.attrib - atttributes of root element

# see XML document
# The tostring() method is used to convert an XML element into a string representation. The encoding parameter is set to 'utf8' to ensure that the output is encoded in UTF-8 format.
# Finally, the decode() method is used to decode the UTF-8 encoded string into a Unicode string, which is then printed to the console.
print(ET.tostring(root, encoding='utf8').decode('utf8'))                           # ET.tostring()


#-----------------------------------------------------------------------------------------------------------------
# You'll see that, for example, the <movie> element contains a couple of "attributes", such as favorite title
# <collection> is the single root element: it contains all the other elements, such as <genre>, or <movie>, which are the child elements or subelements.

# looping in tree structure
for child in root:
    print(child.tag, child.attrib)

# root.iter() will list all subelements under the root that match the element specified. Here, you will list all attributes of the movie element in the tree:
for movie in root.iter('movie'):                                                   # root.iter()
    print(movie.tag, movie.attrib)                                                 # movie.tag, movie.attrib                                       

# Many times elements will not have attributes, they will only have text content. Using the attribute .text, you can print out this content.
for description in root.iter('description'):
    print(description.text)                                                        # description.text


#-----------------------------------------------------------------------------------------------------------------
# XPath expressions - XPath is a query language used to search through an XML quickly. XPath - XML Path Language and uses "path-like" syntax to identify and navigate nodes in an XML document.

# The findall() method is called on the root object with the XPath expression ./genre/decade/movie/[year='1992'].
# This expression selects all movie elements that have a year attribute equal to '1992' and are descendants of a decade element that is a child of a genre element.
# this is search on element
for movie in root.findall("./genre/decade/movie/[year='1992']"):                  # root.findall()
    print(movie.attrib)

# this is search on attributes with @
for movie in root.findall("./genre/decade/movie/format[@multiple='Yes']"):
    print(movie.attrib)
    print(movie.text)
    
# use '...' inside of XPath to return the parent element of the current element.
for movie in root.findall("./genre/decade/movie/format[@multiple='Yes']..."):
    print(movie.attrib)


#-----------------------------------------------------------------------------------------------------------------
# Modifying an XML and writing new XML document
elem_to_replace = root.find("./genre/decade/movie[@title='Back 2 the Future']")    # root.find()
elem_to_replace.attrib['title'] = "Back to the Future"

tree.write("movies_1.1.xml")                                                       # tree.write()


#-----------------------------------------------------------------------------------------------------------------
# Moving Elements
# The two years that are in the wrong decade are the movies from the 2000s. Figure out what those movies are using an XPath expression.
for movie in root.findall("./genre/decade/movie/[year='2000']"):
    print(movie.attrib)

# to create new element 'decade' with 
action = root.find("./genre[@category = 'Action']")
new_dec = ET.SubElement(action,'decade')                                           # ET.SubElement()
new_dec.attrib['years'] = '2000s'

x_men = root.find("./genre/decade/movie[@title='X-Men']")
dec_2000s = root.find("./genre[@category='Action']/decade[@years='2000s']")
dec_1990s = root.find("./genre[@category='Action']/decade[@years='1990s']")
dec_2000s.append(x_men)                                                            # append()
dec_1990s.remove(x_men)                                                            # remove()

print(ET.tostring(root, encoding='utf8').decode('utf8'))  
```

```python
# Serialization - high-level data structures such as lists, tuples,sets, array,dataframe -these objects need to be converted into a sequence of bytes that the computer can understand.
# The next time we want to access the same data structure, this sequence of bytes must be converted back into the high-level object in a process known as deserialization.
# Python’s Pickle module is a popular format used to serialize and deserialize data types.
# The Pickle dump() and dumps() functions are used to serialize an object. The only difference between them is that dump() writes the data to a file, while dumps() represents it as a byte object. 
# Similarly, load() reads pickled objects from a file, whereas loads() deserializes them from a bytes-like object.
# Preserves Python objects : Can save complex objects such as lists, dictionaries, sets, tuples, and custom class instances


import numpy as np
import pickle
import pandas as pd

numpy_array = np.ones([10,10])

# serialization
with open('my_array.pkl', 'wb') as f:
    pickle.dump(numpy_array,f)                                              # pickle.dump()

# deserialization
with open('my_array.pkl', 'rb') as f:
    unpickled_array = pickle.load(f)                                        # pickle.load()

#-----------------------------------------------------------------------------------------------------------------
# panda dataframe
np.random.seed(123)

data = {'Column1': np.random.randint(0, 10, size=100000),
        'Column2': np.random.choice(['A', 'B', 'C'], size=100000),
        'Column3': np.random.rand(100000)}

df = pd.DataFrame(data)

with open('my_df.pkl','wb') as f:
    pickle.dump(df,f)

with open('my_df.pkl','rb') as f:
    unpickled_df = pickle.load(f)

b = pickle.dumps(df)                                                         # pickle.dumps()
unpickled_df = pickle.loads(b)                                               # pickle.loads()
```

# Python debugging in VSCode

```python
# Create a file named var_debug.py in your VS Code project.
# Suppose a trading desk has three positions. You have their quantities, market prices and daily price changes. You want to calculate the portfolio P&L and a simplified historical VaR.
# For this example, we will deliberately introduce a bug in the P&L calculation and use the debugger to find it.



# The bug is this line: "position_pnl = abs(quantity) * price_change" , It discards the sign of the position quantity. For a short position, a positive price change should generate a loss, 
# but abs() makes the calculated P&L positive. The correct line will be "position_pnl = quantity * price_change"


# var_debug.py
def calculate_position_pnl(quantity, price_change):
    """Calculate P&L for one position."""
    position_pnl = abs(quantity) * price_change                   # bug in this line
    return position_pnl


def calculate_portfolio_pnl(positions, price_changes):
    """Calculate total P&L for the portfolio."""
    total_pnl = 0.0

    for i in range(len(positions)):
        quantity = positions[i]
        change = price_changes[i]

        position_pnl = calculate_position_pnl(quantity, change)
        total_pnl += position_pnl

        print(
            f"Position {i}: "
            f"quantity={quantity}, "
            f"change={change}, "
            f"P&L={position_pnl}"
        )

    return total_pnl


def calculate_historical_var(pnl_scenarios, confidence=0.95):
    """Simplified historical VaR from scenario P&Ls."""
    sorted_pnl = sorted(pnl_scenarios)

    index = int((1 - confidence) * len(sorted_pnl))
    var = -sorted_pnl[index]

    return var


def main():
    positions = [100, -50, 80]
    price_changes = [-2, 3, -1]

    portfolio_pnl = calculate_portfolio_pnl(positions, price_changes)
    print(f"\nPortfolio P&L: {portfolio_pnl:.2f}")
    historical_scenarios = [-500, -300, -200, -100, 50,100, -400, 200, -150, 80]
    var_95 = calculate_historical_var(historical_scenarios, 0.95)
    print(f"95% historical VaR: {var_95:.2f}")


if __name__ == "__main__":
    main()
```

**Debugging steps**

- put a breakpoint in line "def calculate_portfolio_pnl(positions, price_changes):"
- Press F5 or select Run → Start Debugging. If prompted, choose Python Debugger and the Python File configuration. VS Code will run your program and pause at the breakpoint.

**Watch panel**

- The Watch panel lets you monitor expressions throughout the debugging session. 
- For example, add these expressions: quantity * change, abs(quantity) * change,position_pnl total_pnl and you can watch them.

**Variables panel**

The Variables panel displays the values of variables in the current execution scope.

For example:
- Local: variables belonging to the current function, such as quantity and position_pnl.
- Global: variables defined at the module level.
- Closure: variables captured from an enclosing function, when applicable.
You can expand lists and dictionaries to inspect their contents. This is useful when checking large arrays, market data and configuration objects.

**Debug Console**

The Debug Console lets you evaluate Python expressions while execution is paused.

For example, enter: <quantity * change> The console is useful for testing a calculation without editing the source file.
Be careful: expressions that modify variables or mutate objects can change the running program's state. Use read-only expressions when you just want to inspect values.

**Call Stack**

The Call Stack tells you how the program reached the current function.


### Do you need a launch.json file?

```python
# Not necessarily. For a simple Python file, select the file and press F5. VS Code can use its default Python debugging configuration.
# A launch.json file becomes useful when you need to specify command-line arguments, a working directory, environment variables or a module to run.
# For example, if your program is main.py and needs to run from the 20260930 directory, a configuration could look like this:

# JSON file
{
    "version": "0.2.0",
    "configurations": [
        {
            "name": "Debug Market Risk",
            "type": "debugpy",
            "request": "launch",
            "program": "${workspaceFolder}/20260930/main.py",
            "cwd": "${workspaceFolder}/20260930",
            "console": "integratedTerminal",
            "justMyCode": true
        }
    ]
}

# This assumes your VS Code workspace is the Python Implementation directory and main.py is the entry point.
# To create the file, open Run and Debug → create a launch.json file, select the Python debugger, and edit the generated configuration.
```

| Setting | Meaning |
|---|---|
| `name` | Name shown in the debug configuration selector. |
| `type` | Debugger type; `debugpy` is used for Python debugging. |
| `request` | `launch` starts a new program; `attach` connects to an existing process. |
| `program` | Python script to execute. |
| `cwd` | Working directory for the process. |
| `console` | Where program input and output appear. |
| `justMyCode` | When `true`, generally avoids stepping into non-user/library code. |
