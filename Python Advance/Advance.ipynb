### BSM Model

```python
import scipy.stats as scs
Image.open('C:\\Users\\Admin\\Python Advance\\BSM.JPG')
```

```python
s = 100
k= 100
r = 0.02
sigma = 0.05
t=1

def bsm(s,k,r,t,sigma,type = 'c'):
    d1 = np.log(s/k) + (r+sigma**2/2)*t
    d2 = d1 - sigma*np.sqrt(t)
    nd1 = scs.norm.cdf(d1)
    nd1_neg = scs.norm(-d1)
    nd2 = scs.norm.cdf(d2)
    nd2_neg = scs.norm(-d2)
    
    try:
        if type == 'c':
            price = nd1*s - nd2*k*np.exp(-r*t)
        elif type == 'p':
            price = k*np.exp(-r*t)*nd2_neg -s*nd1_neg
    except:
        print('please check all input parameters')
        
    else:
        print('input parameters are good')
        
    return price


bsm(100,100,0.02,1,0.05,'c')
```

### Difference between NaN and None

The distinction between None and NaN in Pandas can be summarized as:

- None represents a missing entry, but its type is not numeric. So any column (ad Pandas Series) that contains a None value is definately not a numeric type, such as int or float.
- NaN which stands for not-a-number, is on the other hand a numeric type. This means that NaN can be found in a numeric column of int or float type.

### Part A(Process class in multiprocessing)

A program is CPU bound if it would go faster if the CPU were faster, i.e. it spends the majority of its time simply using the CPU (doing calculations). 
A program that computes new digits of π will typically be CPU-bound, it's just crunching numbers.

A program is I/O bound if it would go faster if the I/O subsystem was faster. 
Which exact I/O system is meant can vary; I typically associate it with the disk, but of course, networking or communication, 
in general, is common too. A program that looks through a huge file for some data might become I/O bound since the bottleneck is then the reading of the data from disk 

A CPU-bound process requires more CPU time or spends more time in the running state . An I/O-bound process requires more I/O time and less CPU time. An I/O-bound process spends more time in the waiting state.

A general rule of thumb- you use threads for things which are IO bound, and processes fror things which are CPU bound.

### Part B(Process class in multiprocessing)

```python
import multiprocessing
import time
import os


start = time.perf_counter()                                        # perf_counter() -Performance counter for benchmarking

# here _ is throwaway variable, means no informaton is rquired for the same, used only for looping 

def do_something(secs):
    print(f"sleep for {secs} seconds")
    print('module name:', __name__)                                   # __name__
    print('parent process:', os.getppid())                            # os.getppid() - parent process id
    print('process id:', os.getpid())                                 # os.getpid() - process id, if 2 cores/process used then 2 process ids 
    time.sleep(secs)
    print(f"done sleeping for {secs} seconds")
    
processes = []
for _ in range(10):
    p= multiprocessing.Process(target=do_something, args=[1.5])         # multiprocessing.Process() - process object
    p.start()                                                           # p.start()
    processes.append(p)                                                 # append(p)

for process in processes:
    process.join()                                                      # procss.join()
    

finish = time.perf_counter()
print(f"finished in {finish-start} seconds")
```

### Part C(concurrent.futures) - multiprocessing and multithreading

```python
import concurrent.futures
import os

def unit_task(num):
    print('module name:', __name__)                                   # __name__
    print('parent process:', os.getppid())                            # os.getppid() - parent process id
    print('process id:', os.getpid())                                 # os.getpid() - process id, if 2 cores/process used then 2 process ids 
    return num

# this is for multiprocessing

def main():
    
    # context manager of ProcessPoolExecutor object
    # advantae of context manager - it will automaticaly join all processes once these are completed and only then step over to next line.
    # multiprocessing.Pool.map() method returns a list while the concurrent.futures.ProcessPoolExecutor.map() method returns an iterator just like the built-in map method.
    # normally we prefer map() method
    
    with concurrent.futures.ProcessPoolExecutor() as executor:                           # concurrent.futures.ProcessPoolExecutor()
        results = [executor.submit(unit_task, i) for i in range(10)]                     # executor.submit()

    # print the results of executors as and when they are completed

    for result in concurrent.futures.as_completed(results):                              # concurrent.futures.as_completed()
        print(result.result())                                                           # result.results()                                                

    # OR

    with concurrent.futures.ProcessPoolExecutor() as executor:                           
        results = executor.map(unit_task, (1,2,3,4,5))   

    for result in results:
        print(result)


# this is for multithreading

def main():
    
    # context manager of ThreadPoolExecutor object
    # advantae of context manager - it will automaticaly join all threds once these are completed and only then step over to next line.


    with concurrent.futures.ThreadPoolExecutor() as executor:                           
        results = executor.map(unit_task, (1,2,3,4,5))   

    for result in results:
        print(result)

if __name__ == "__main__":
    main()
```

### Part D(Pool class in multiprocessing)

```python
from multiprocessing import Pool
import os

def unit_task(num):
    print('module name:', __name__)                                   # __name__
    print('parent process:', os.getppid())                            # os.getppid() - parent process id
    print('process id:', os.getpid())                                 # os.getpid() - process id, if 2 cores/process used then 2 process ids 
    return num

os.cpu_count()                                                        # os.cpu_count() - # of cores available in system

def main():
    
    p = Pool(processes=4)                                             # Pool() if not defined then it will be # of cores in CPU
    results = p.map(func = unit_task, iterable= range(10))            # map()
    p.close()                                                         # close()
    print(results)

    # OR
    
    # multiprocessing.Pool objects as context managers
    # multiprocessing.Pool.map() method returns a list while the concurrent.futures.ProcessPoolExecutor.map() method returns an iterator just like the built-in map method.
    
    with Pool(processes=4) as p:
        results = p.map(func = unit_task, iterable= range(10))            
        p.close()                                                         
        print(results)
        
if __name__== "__main__":
    main()
```
